```python
df['Fraud_flag']=df.apply(lambda row: 1 if (row['Annual_Income']<500000 and row['Num_of_Loan']>7)else 0,axis=1)
```

```python
from sklearn.ensemble import IsolationForest #ML algorithem for anomoly detection
```

```python
x=df[['Annual_Income','Num_of_Loan']]
model=IsolationForest(contamination=0.05,random_state=42)
df['fraud_flag']=model.fit_predict(x)
df['fraud_flag']=df['fraud_flag'].map({1:0,-1:1})
```

```python
sns.scatterplot(x='Annual_Income',y='Num_of_Loan',hue='fraud_flag',data=df,palette={0:'green',1:'red'}) #showing fraud vs non fraud flag
plt.title('Detected Fraud vs Non-fraud')
plt.show()
```
